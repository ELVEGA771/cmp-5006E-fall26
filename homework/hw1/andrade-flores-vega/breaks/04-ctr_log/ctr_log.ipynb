## Execution Example
This notebook contains a series of crib-dragging executions against the audit log. Every entry was encrypted with the same key **and the same nonce**, so they share one keystream and `C_i ⊕ C_j = P_i ⊕ P_j`. The attacker only knows the log format (`YYYY-MM-DD HH:MM user=... action=... result=... from=IP`) and goes back and forth until `log2` is recovered.

```python
from ctr_log import crib_drag, recover_entry
from duel1_targets import ctr_log_entries, _xor
```

```python
ciphertexts: dict[str, str] = ctr_log_entries()
ciphertexts_bytes: dict[str, bytes] = {
    key: bytes.fromhex(value) for key, value in ciphertexts.items()
}

for key, value in ciphertexts_bytes.items():
    print(key, len(value), "bytes")
```

Every log line starts with a timestamp, so the date is the first crib. Offset 0 is readable: all entries share the same date and hour.

```python
crib_drag(_xor(ciphertexts_bytes["log0"], ciphertexts_bytes["log2"]), "2025-03-01 12:0")
```

The format says a `user=` field follows the timestamp. Guessing that the target line is an admin action, we drag `user=admin` over `log0 ⊕ log2`: offset 17 reveals who wrote `log0`.

```python
crib_drag(_xor(ciphertexts_bytes["log0"], ciphertexts_bytes["log2"]), "user=admin")
```

`log0` belongs to `alice`. A user's most common event is a login, so we extend the crib with `user=alice action=login` and get the user and action of `log2`.

```python
crib_drag(_xor(ciphertexts_bytes["log0"], ciphertexts_bytes["log2"]), "user=alice action=login")
```

`log2` is `user=admin action=expor...`. Now we guess the whole `log0` line (successful login from an internal IP) and recover `log2` directly.

```python
known_log0 = "2025-03-01 12:00 user=alice action=login result=success from=10.0.0.5"

recover_entry(ciphertexts_bytes["log0"], ciphertexts_bytes["log2"], known_log0)
```

**Confirmation:** if our guess of `log0` were wrong, it would produce garbage against the *third* ciphertext too. Using the same guess against `log1` gives a perfectly valid log line, so the guess (and therefore the recovered `log2`) is correct.

```python
recover_entry(ciphertexts_bytes["log0"], ciphertexts_bytes["log1"], known_log0)
```

`log2` is one byte longer than the other entries, and that last keystream byte was never used by any other entry, so XOR cannot reveal it. We can only say it is the last digit of the IP `10.0.0.?`.

```python
recovered_log2 = recover_entry(ciphertexts_bytes["log0"], ciphertexts_bytes["log2"], known_log0)

print("recovered bytes:", len(recovered_log2), "of", len(ciphertexts_bytes["log2"]))
print(recovered_log2.decode() + "?")
print("contains 'user=admin action=export':", b"user=admin action=export" in recovered_log2)
```

**Answer of `log2`**: 2025-03-01 12:09 user=admin action=export result=success from=10.0.0.? (the last digit of the IP is not recoverable from the other entries)
